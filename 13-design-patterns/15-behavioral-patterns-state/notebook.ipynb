# State Pattern

## Overview

**Category:** Behavioral

The State pattern allows an object to **alter its behaviour when its internal state
changes**. The object appears to change its class. Instead of a large switch statement
inside the context object, each state is represented by its own class that handles
the behaviour appropriate for that state.

## Key Concepts

- **Context** (`ShoppingCart`) — maintains a reference to the current state object; delegates state-specific behaviour to it
- **State ABC** (`AbsState`) — defines the interface for all concrete states
- **Concrete State** — implements the behaviour for one specific state (Empty, NotEmpty, AtCheckout, PaidFor)

### Motivating Example: Shopping Cart
A shopping cart behaves differently depending on its state:
- **Empty** — can add items; cannot checkout or remove items
- **Not Empty** — can add or remove items; can proceed to checkout
- **At Checkout** — can remove items; can pay; cannot add items
- **Paid For** — terminal state

Transitions are delegated to the state objects themselves, keeping transition logic
co-located with state behaviour.

### Consequences
- Encapsulates state-specific behaviour in dedicated classes
- Makes state transitions explicit and localised
- State objects can be shared (flyweight states)
- Eliminates large conditional blocks in the context
