# Adapter Pattern

## Overview

**Category:** Structural

The Adapter pattern converts the interface of a class into another interface that clients
expect. It lets incompatible classes work together without modifying their source code —
like a power adapter that lets a US plug fit a European socket.

## Key Concepts

### The Problem
You have a `Customer` object with an `address` property (full address as a string), but a
vendor API expects separate `number` and `street` properties. You need them to work together
without changing either side.

### Classification: Structural
- Converts the interface of a class into one that clients expect
- Lets incompatible classes work together
- Can provide additional functionality during adaptation

### Two Types of Adapters

**Object Adapter (Composition)**
- Wraps the adaptee in a new class
- Delegates calls to the wrapped object
- Works with all subclasses of the adaptee
- Preferred in Python (composition over inheritance)

**Class Adapter (Multiple Inheritance)**
- Inherits from both the target interface and the adaptee
- Can override adaptee methods
- Tied to one specific adaptee subclass

### Pros and Cons

| | Object Adapter | Class Adapter |
|---|---|---|
| Mechanism | Composition | Multiple inheritance |
| Flexibility | Works with all adaptee subclasses | Committed to one subclass |
| Method override | Via delegation | Direct override |

## Demos

### Before: Incompatible Interfaces

Original program prints customer addresses. Vendor objects have a different interface. See how client code breaks when mixing the two.

### Implementation: Object Adapter

A wrapper class (`VendorAdapter`) holds a reference to the vendor object and exposes the `address` property that the client expects.

### Implementation: Class Adapter

The adapter subclasses the vendor class and adds the `address` property — using Python's multiple inheritance.
