# Factory & Abstract Factory Pattern

## Overview

**Category:** Creational

The Factory and Abstract Factory patterns decouple object creation from usage, so client
code never needs to know which concrete class it is instantiating. The **Simple Factory**
centralizes conditional logic in one place; the **Factory Method** delegates creation to
subclasses; the **Abstract Factory** provides an interface for creating *families* of
related objects without specifying concrete classes.

## Key Concepts

### Simple Factory
- A single factory class with a method that returns one of several possible objects
- Uses a dictionary (or dynamic loading) to map names → classes
- Client calls `factory.create_instance(name)` — no `if/elif` chains in client code

### Factory Method Pattern
- Defines an abstract `create_product()` method in a base factory class
- Each concrete factory subclass implements `create_product()` for one product type
- Client works only with the abstract factory interface

### Abstract Factory Pattern
- Provides an interface for creating *families* of related objects (e.g. economy, sport, luxury cars)
- Concrete factories implement the interface for a specific product family (Ford vs GM)
- Guarantees that products from the same factory are compatible

### When to Use
- When the exact type of object to create isn't known until runtime
- When you want to centralize and encapsulate object creation logic
- When you need to swap entire families of products (Abstract Factory)

## Demos

### Before: Hardcoded Object Creation (no factory)

Client code uses `if/elif` chains to instantiate the right car class directly. Adding a new car type requires editing the client.

### Simple Factory: Using a Dictionary-Based Factory

An `AutoFactory` class loads all available car classes at startup and creates instances by name — no conditionals in client code.

### Implementation: Full Factory Method Pattern

Each brand gets its own factory class. A dynamic loader finds and instantiates the right factory. Client code is completely decoupled from concrete car classes.

### Implementation: Abstract Factory Pattern

Factories now create entire *families* of cars (economy / sport / luxury). Swapping `FordFactory` for `GMFactory` changes the entire product lineup.
