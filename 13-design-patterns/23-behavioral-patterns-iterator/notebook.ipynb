# Iterator Pattern

## Overview

**Category:** Behavioral (also known as the **Cursor** pattern)

The Iterator pattern provides a way to **sequentially access elements of a collection**
without exposing its underlying representation. Python has first-class iterator support
built into the language — making this pattern both natural and powerful in Python.

## Key Concepts

- **Iterator** — knows how to traverse a collection; maintains traversal state
- **Iterable** — the collection that can produce an iterator
- **`__iter__`** — returns the iterator object
- **`__next__`** — returns the next element; raises `StopIteration` when exhausted

### Two Python Iterator Protocols

| Protocol | Method | Use case |
|---|---|---|
| **Sequence** | `__getitem__(index)` | Collections with integer indexing |
| **Callable/Iterator** | `__iter__()` + `__next__()` | Any traversable collection |

### Python Generators
Python generators (`yield`) provide the simplest way to implement iterators. A generator
function automatically implements the iterator protocol — no class needed.

```python
def employee_generator(employees):
    for emp in employees:
        yield emp
```

### When to Use
- When you need a uniform way to traverse different collection types
- When you want multiple simultaneous traversals
- When the collection implementation might change
