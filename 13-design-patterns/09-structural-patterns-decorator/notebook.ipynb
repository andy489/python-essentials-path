# Decorator Pattern

## Overview

**Category:** Structural (also known as the **Wrapper** pattern)

The Decorator pattern attaches additional responsibilities to an object *dynamically* at
runtime — a flexible alternative to subclassing for extending functionality. Decorators
wrap an existing object and add behaviour before or after delegating to the wrapped object.

## Key Concepts

- **Component** (`AbsCar`) — the interface shared by concrete objects and decorators
- **Concrete Component** — the base object being decorated (e.g. `EconomyCar`)
- **Abstract Decorator** — holds a reference to a component and delegates to it
- **Concrete Decorator** — adds one specific responsibility (e.g. `V6Engine`, `LeatherUpholstery`)

### The Problem Without Decorator
A car can have:
- Engine: 4cyl or 6cyl
- Paint: white, red, black
- Upholstery: leather or vinyl

That's 2 × 3 × 2 = **12 subclasses** for one base model. With three base models: **36 classes**.
The subclassing approach explodes.

### Advantages
- No practical limit on combinations — stack as many decorators as needed
- Transparent to clients — decorators and components share the same interface
- Follows Open/Closed principle — add features without modifying existing code

### Python Note
Python has built-in decorator syntax (`@decorator`). The GoF Decorator pattern is a
*structural design pattern* that operates on objects. Python's `@property` syntax is
related but distinct.
