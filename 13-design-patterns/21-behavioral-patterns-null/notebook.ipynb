# Null Object Pattern

## Overview

**Category:** Behavioral (sometimes classified as Creational)

The Null Object pattern provides a **default object with do-nothing behaviour** as a
substitute for `None`. Clients never need to check whether the object they received is
valid — the null object simply does nothing gracefully, eliminating defensive null checks
throughout client code.

## Key Concepts

- Implement the same interface as the real object
- All methods do nothing (or return safe defaults)
- Return the null object from factories when no real object matches

### When to Use
- Wherever you find yourself writing `if obj is not None:` repeatedly
- When a missing object should behave like a real object (just doing nothing)
- Creational patterns (Factory, Abstract Factory) benefit from a null object as a default return

### Example

```python
# Without Null Object
cust = factory.get_customer(id)
if cust is not None:
    cust.send_invoice()

# With Null Object — the null customer's send_invoice() does nothing
cust = factory.get_customer(id)
cust.send_invoice()
```

### Consequences
- Eliminates null checks in client code
- Clients are simpler and more readable
- Behaviour for the "missing" case is centralised in one class
- Can mask programming errors if overused

## Demos

### Before: Manual Null Checks

Factory returns `None` for unknown types. Client code must check before every call — tedious and error-prone.

### Implementation: Null Object Pattern

Factory returns a `NullClass` object instead of `None`. Client code calls methods unconditionally — the null object absorbs all calls silently.
