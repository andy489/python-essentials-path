# Builder Pattern

## Overview

**Category:** Creational

The Builder pattern separates the *construction* of a complex object from its
*representation*, so the same construction process can create different representations.
It is especially useful when an object requires many optional parameters or a multi-step
assembly process — avoiding telescoping constructors and scattered attribute assignments.

## Key Concepts

- **Director** — orchestrates the build steps in the correct order
- **Abstract Builder** — defines the interface for each build step
- **Concrete Builder** — implements the steps for a specific variant of the product
- **Product** — the complex object being assembled

### When to Use
- When constructing a complex object step by step
- When the same construction process should produce different results
- When you want to isolate complex construction code from business logic

### Evolution Shown in Demos
1. **BeforeBuilder1** — pass all args to constructor (unwieldy)
2. **BeforeBuilder2** — set attributes after construction (fragile)
3. **BeforeBuilder3** — move construction into a separate class (better, no interface)
4. **BeforeBuilder4** — add build steps as methods (almost there)
5. **Builder** — introduce `Director` + `AbsBuilder` + multiple concrete builders

## Demos

### Before 1: Constructor with All Parameters

All six component choices passed directly to `__init__`. Hard to read, easy to mix up argument order.

### Before 2: Set Attributes After Construction

Use a no-arg constructor and set each attribute explicitly. No enforcement that all parts are set.

### Before 3: Dedicated Builder Object (no interface)

Construction logic moves into `MyComputer`. Cleaner, but no abstraction — can't swap builders.

### Before 4: Build Steps as Methods

Builder now exposes discrete step methods. Still no abstract interface, so the Director can't work with different builders.

### Implementation: Full Builder Pattern

A `Director` drives `AbsBuilder`. Swap `MyComputerBuilder` for `BudgetBoxBuilder` to produce a different computer — same process, different result.
