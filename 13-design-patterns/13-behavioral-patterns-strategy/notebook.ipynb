# Strategy Pattern

## Overview

**Category:** Behavioral (also known as the **Policy** pattern)

The Strategy pattern defines a **family of algorithms**, encapsulates each one, and makes
them interchangeable. The algorithm used by a client can be swapped at runtime without
changing the client itself.

## Key Concepts

- **Context** — holds a reference to a strategy; delegates algorithm execution to it
- **Strategy ABC** — defines the common interface all concrete strategies must implement
- **Concrete Strategy** — one specific algorithm implementation (e.g. FedEx, UPS, Postal)

### When to Use
- Multiple related classes differ only in their behaviour
- You need different variants of an algorithm
- An algorithm uses data the client shouldn't know about
- A class defines many behaviours via conditionals — replace each branch with a strategy

### Python Variations
Besides the classic class-based approach, Python supports:
- **Function strategies** — pass a plain function as the strategy
- **Lambda strategies** — inline one-liners for simple algorithms

These lightweight variants eliminate the need for a separate ABC when the algorithm is simple.

## Demos

### Before: Hardcoded Shipping Logic

An `Order` class uses `if/elif` chains to calculate shipping cost per carrier. Adding a new carrier means editing the `Order` class.

### Implementation: Strategy Pattern

Each carrier's shipping calculation is extracted into its own strategy class. `Order` holds a reference to a strategy — swap the strategy to change behaviour.

### Variation: Functions and Lambdas as Strategies

For simple algorithms, Python functions and lambdas can serve as strategies without the overhead of a full class hierarchy.
