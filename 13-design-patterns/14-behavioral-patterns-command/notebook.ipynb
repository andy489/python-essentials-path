# Command Pattern

## Overview

**Category:** Behavioral

The Command pattern encapsulates a **request as an object**, allowing requests to be
parameterised, queued, logged, and undone. It decouples the object that *sends* a request
from the objects that *receive and process* it.

## Key Concepts

- **Command** (`CommandABC`) — encapsulates a request as an object with an `execute()` method
- **Concrete Command** — implements one specific operation (CreateOrder, UpdateQuantity, ShipOrder)
- **Invoker** — asks the command to carry out a request
- **Receiver** — knows how to perform the work
- **Client** — creates and configures command objects

### Motivating Example: Command-Line Order Processing
A program accepts command-line arguments to create, update, and ship orders:
```
python __main__.py CreateOrder ...
python __main__.py UpdateQuantity ...
python __main__.py ShipOrder ...
```

The Command pattern maps each argument to a command object — adding new operations
requires only a new command class, not changes to the invoker.

### Undo Support
By adding an `undo()` method to each command and maintaining a history stack, full
undo/redo functionality is straightforward — a key benefit of the pattern.

### Before: Direct Command Dispatch

A `CommandExecutor` uses `if/elif` chains keyed on the command name. Adding a new command requires editing the executor.

### Implementation: Command Pattern

Each operation is its own command class. The invoker only knows the `execute()` interface — new commands require zero changes to existing code.

### Extension: Undo Support

Each command implements both `execute()` and `undo()`. A history stack enables full undo/redo for home automation actions.
