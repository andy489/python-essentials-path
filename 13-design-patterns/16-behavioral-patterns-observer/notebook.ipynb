# Observer Pattern

## Overview

**Category:** Behavioral (also known as **Dependents** or **Publish-Subscribe**)

The Observer pattern defines a **one-to-many dependency** between objects: when one object
(the Subject) changes state, all its dependents (Observers) are notified and updated
automatically. It is the foundation of event-driven programming and the MVC architecture.

## Key Concepts

- **Subject** (`SubjectABC`) — maintains a list of observers; notifies them on state change
- **Observer** (`ObserverABC`) — defines the update interface
- **Concrete Subject** — stores state; sends notifications when state changes
- **Concrete Observer** — implements the update logic (e.g. `CurrentKPIs`, `ForecastKPIs`)

### When to Use
- When a change in one object requires changing others, and you don't know how many objects need to change
- When an object should notify other objects without assumptions about who those objects are

### Dangling Reference Bug
Python uses reference counting for garbage collection. If observers are held in a regular
`set`, the subject keeps a reference alive — observers are never collected even after going
out of scope. The fix: use **weak references** (`weakref`) or **context managers** to
ensure observers detach cleanly.

## Demos

### Before: Direct KPI Update Calls

KPI displays are updated by direct method calls scattered through the business logic. Adding a new display requires editing business code.

### Implementation: Observer Pattern

Subject notifies all registered observers automatically. New displays register themselves — business logic is untouched.

### Fix: Context Manager Observers

Solves the dangling reference bug: observers register on `__enter__` and detach on `__exit__`, ensuring proper garbage collection.
