# Working with Regular Expressions in Python

A **regular expression** (regex) is a sequence of characters that represents a text pattern. It is used to:
- Search for patterns in text
- Validate data formats
- Replace or transform text

Python's built-in `re` module provides all the tools you need.

```python
import re
```

## Reading Regex Left to Right

A regex pattern is always read **left to right** — just like reading text. Each character (or special symbol) has a meaning.

Regex is **case-sensitive** by default — `sample` and `Sample` are different patterns.

## Literal Character Matching

The simplest patterns match characters literally.

```python
# example_search.py — literal match
text = "This is a sample text."
pattern = "sample"

match = re.search(pattern, text)

if match:
    print("Found a match!")
else:
    print("No match found.")

# Case-sensitive: 'Sample' won't match
print(re.search("Sample", text))  # None
```

## The Wildcard `.`

The dot `.` matches **any single character** except a newline.

```python
# f.n matches 'fun', 'fan', 'fin', etc.
print(re.search(r"f.n", "Python regex is fun."))  # matches 'fun'
print(re.search(r"f.n", "I'm a fan."))             # matches 'fan'
print(re.search(r"f..n", "I'm a fan."))            # no match — needs 2 chars between f and n
```

## Anchors: `^` and `$`

Anchors do not match characters — they match **positions**:
- `^` — start of the string
- `$` — end of the string

```python
# example_match.py — re.match only matches at the start of the string
pattern = r'The'
text = 'The quick brown fox'

match = re.match(pattern, text)

if match:
    print(f'The text string starts with the word "The"')
else:
    print(f'The text string does not start with the word "The"')

# Anchors with re.search
text2 = "Pluralsight"
print(re.search(r"^sight", text2))   # no match — 'sight' is not at the start
print(re.search(r"^Plural", text2))  # match — 'Plural' is at the start
print(re.search(r"sight$", text2))   # match — 'sight' is at the end
```

## Escape Character `\` and `re.escape`

To match a special character **literally**, prefix it with `\`. Use `re.escape()` to automatically escape all special characters in a string.

```python
# Manual escaping
print(re.search(r"regex\.", "I like regex."))  # matches 'regex.'
print(re.search(r"regex\.", "I like regex!"))  # no match — '!' ≠ '.'

# example_escape.py — re.escape automatically escapes special characters
pattern = r'.'  # in regex: matches ANY character
text = 'The quick brown fox.'

escaped_pattern = re.escape(pattern)
print("new pattern: ", escaped_pattern)  # \. — now matches a literal dot

match = re.search(escaped_pattern, text)
if match:
    print(f'Found a dot in the text')
else:
    print(f'Could not find dot in the text')
```

## Character Classes `[...]`

A character class (or character set) matches **one character** from a defined group.

- `[ae]` — matches 'a' or 'e'
- `[a-z]` — matches any lowercase letter
- `[a-g]` — matches letters a through g
- `[a-gA-G]` — matches a–g or A–G

Inside `[...]`, only `]`, `\`, `^`, and `-` have special meaning.

```python
# Match both British and American spellings
print(re.search(r"apologi[sz]e", "I'd like to apologise."))   # match
print(re.search(r"apologi[sz]e", "I'd like to apologize."))   # match
```

## Negating a Character Class `[^...]`

A `^` inside `[...]` at the start **negates** the class — matches any character NOT in the set.

```python
# Match any character that is not a, b, or c
print(re.findall(r"[^abc]", "abcdef"))  # ['d', 'e', 'f']
```

## Quantifiers

Quantifiers specify how many times a pattern should repeat:

| Quantifier | Meaning |
|---|---|
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 (optional) |
| `{n}` | exactly n times |
| `{n,m}` | between n and m times |

```python
print(re.findall(r"\d+", "Order 42 and 100 items"))   # ['42', '100']
print(re.findall(r"\d{3}", "Phone: 123-456-7890"))     # ['123', '456', '789']
print(re.findall(r"colou?r", "color colour"))           # ['color', 'colour']
```

## Grouping `(...)`

Parentheses group parts of a pattern and capture matched text.

```python
match = re.search(r"(\d{4})-(\d{2})-(\d{2})", "Date: 2024-01-15")
if match:
    print("Full match:", match.group(0))
    print("Year:",  match.group(1))
    print("Month:", match.group(2))
    print("Day:",   match.group(3))
```

## Common `re` Module Functions

| Function | Description |
|---|---|
| `re.search(pattern, text)` | Find first match anywhere in text |
| `re.match(pattern, text)` | Match only at the beginning |
| `re.fullmatch(pattern, text)` | Match must cover the whole string |
| `re.findall(pattern, text)` | Return all non-overlapping matches |
| `re.sub(pattern, repl, text)` | Replace matches with a string |
| `re.split(pattern, text)` | Split string by matches |
| `re.compile(pattern)` | Compile pattern for reuse |

```python
text = "The price is $42 and $100"

# search — first match
print(re.search(r"\$\d+", text).group())        # $42

# findall — all matches
print(re.findall(r"\$\d+", text))               # ['$42', '$100']

# sub — replace
print(re.sub(r"\$\d+", "[PRICE]", text))        # replace prices

# split — split on whitespace runs
print(re.split(r"\s+", "one  two   three"))     # ['one', 'two', 'three']
```

## Raw Strings

Always use raw strings (`r"..."`) for regex patterns. This prevents Python from interpreting backslashes before the regex engine sees them.

```python
# Without r prefix: \b would be interpreted as backspace by Python
pattern = re.compile(r"\bword\b")
print(pattern.findall("a word is a word"))  # ['word', 'word']
```
