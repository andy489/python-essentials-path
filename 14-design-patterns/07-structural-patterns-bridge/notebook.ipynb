# Bridge Pattern

## Overview

**Category:** Structural

> "Decouple an abstraction from its implementation so that the two can vary independently."
> — *Design Patterns: Elements of Reusable Object-Oriented Software*, GoF (1994)

The Bridge pattern splits a large class (or set of closely related classes) into two
separate hierarchies — **abstraction** and **implementation** — that can be developed and
extended independently.

## Key Concepts

### The Problem Without Bridge
A subscription system has `Monthly` and `Annual` plans. When discount types (corporate,
student) are added, the class count explodes:
- `MonthlySubscription`
- `AnnualSubscription`
- `MonthlyCorporateSubscription`
- `MonthlyStudentSubscription`
- `AnnualCorporateSubscription`
- `AnnualStudentSubscription`
- ... and more as features grow

### The Solution
Bridge extracts the discount logic into its own hierarchy. The subscription holds a
**reference** to a discount object — composition replaces inheritance.

### Key Roles
- **Abstraction** (`Subscription`) — high-level control, holds reference to implementor
- **Refined Abstraction** — extends abstraction with more specific behaviour
- **Implementor** (`Discount`) — defines the interface for implementation classes
- **Concrete Implementor** — the actual discount/extension logic

### Benefit
Adding a new discount type = one new class. Adding a new subscription type = one new class.
No combinatorial explosion.

## Demos

### Before 1: Simple Subscriptions (no discounts)

Basic `Monthly` and `Annual` classes with no variation support.

### Before 2: Discounts via Subclassing

Adding corporate and student discounts multiplies the class count. This is the explosion the Bridge pattern solves.

### Implementation: Bridge Pattern with Discounts

Discount logic is extracted into a separate `Discount` hierarchy. Subscriptions hold a discount reference.

### Implementation: Bridge with Special Offers

New requirement: subscriptions can be extended. Add an `Extend` implementor without touching the subscription classes.
