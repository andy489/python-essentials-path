# Mediator Pattern

## Overview

**Category:** Behavioral

The Mediator pattern defines an object that **encapsulates how a set of objects interact**.
It promotes loose coupling by preventing objects from referring to each other explicitly —
all communication goes through the mediator. This reduces tangled peer-to-peer dependencies
and makes the system easier to maintain.

## Key Concepts

- **Mediator** — centralises communication logic between colleagues
- **Colleague** — each object knows only the mediator, not the other colleagues

### The Problem: Object Interaction Explosion
With *n* objects that interact with each other, you have up to *n(n-1)/2* connections.
As the system grows, each object becomes harder to reuse because it drags in all its peers.

### Motivating Example: Pet Management
Three pets (cat, dog, fish) need feeding, walking, and attention at different times of day.
Without a mediator, each pet needs to know about the others' schedules. The mediator
centralises all scheduling decisions.

### Consequences

**Benefits:**
- Reduces subclassing
- Decouples colleagues — increases reusability
- Simplifies object protocols (many-to-many → one-to-many)
- Centralises control

**Drawbacks:**
- The mediator itself can become complex and hard to maintain
- It can become a "God object" that knows too much

## Demos

### Before: Direct Peer-to-Peer Interaction

Pets interact with each other directly, creating tangled dependencies. The fish's feeding depends on whether it's alive — but only the fish knows that.

### Implementation: Mediator Pattern

A `PetMediator` orchestrates all pet interactions. Each pet only calls the mediator — no direct peer references.
