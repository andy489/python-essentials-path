# Composite Pattern

## Overview

**Category:** Structural

The Composite pattern lets you compose objects into **tree structures** to represent
part-whole hierarchies. Clients treat individual objects and compositions of objects
uniformly — the same code works whether it's operating on a single leaf or an entire subtree.

## Key Concepts

- **Component** (`AbsComposite`) — common interface for both leaves and composites
- **Leaf** — a node with no children; implements the component interface directly
- **Composite** — a node with children; implements the component interface by delegating to children
- **Client** — works only with the component interface; unaware whether it has leaves or composites

### Key Insight: Part = Mini Whole
Any part of the tree looks like a smaller version of the whole tree. This **self-similarity**
is the defining property of the Composite pattern.

### Motivating Example: Family Tree
- A `Person` can stand alone (leaf) or be part of a family (composite)
- Finding the oldest person should work the same way whether the tree has 1 node or 1000

### When to Use
- Representing part-whole hierarchies (file systems, UI component trees, org charts)
- When clients should treat individual items and collections identically
