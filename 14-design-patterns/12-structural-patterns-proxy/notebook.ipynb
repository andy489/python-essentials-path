# Proxy Pattern

## Overview

**Category:** Structural

The Proxy pattern provides a **surrogate or placeholder** for another object to control
access to it. The proxy implements the same interface as the real subject, so clients
use it transparently — and the proxy intercepts calls to add access control, caching,
logging, or lazy initialisation.

## Key Concepts

### Four Types of Proxies

| Type | Purpose |
|---|---|
| **Protection Proxy** | Controls access based on permissions |
| **Virtual Proxy** | Delays creation of expensive objects (`@functools.lru_cache`) |
| **Remote Proxy** | Hides communication details (e.g. `pyodbc` for remote DB) |
| **Smart Reference** | Adds reference counting or logging around the real object |

### Key Roles
- **Subject** (`AbsEmployees`) — interface shared by real subject and proxy
- **Real Subject** (`Employees`) — the actual object doing the work
- **Proxy** — controls access, delegates to the real subject when permitted

### Consequences
- Introduces a level of indirection
- Protection proxy adds security without changing the real subject
- Virtual proxy enables lazy instantiation
- Transparent to clients — they use the proxy as if it were the real object
