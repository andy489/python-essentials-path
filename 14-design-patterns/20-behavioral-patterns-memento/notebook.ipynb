# Memento Pattern

## Overview

**Category:** Behavioral

The Memento pattern **captures and externalises an object's internal state** so it can be
restored later, without violating encapsulation. The saved state (the memento) is opaque
to everything except the originator that created it.

## Key Concepts

- **Originator** — the object whose state needs to be saved/restored; creates and uses mementos
- **Memento** — stores the internal state of the originator; only the originator can read it
- **Caretaker** — holds the memento; never examines or modifies its contents

### Motivating Example: Save Game
A hero in a complex game can die. The game should save state at checkpoints and allow
the player to restore it. The Memento pattern implements this "save game" snapshot without
exposing the hero's internal state to the save/load system.

### Consequences

**Benefits:**
- Preserves encapsulation — state is opaque outside the originator
- Simplifies the originator — it doesn't manage its own history
- Easy state restoration

**Drawbacks:**
- Mementos can be expensive if the state is large
- Caretaker must track memento lifecycle
- Python pickles are a natural fit for the Memento pattern

## Demos

### Before: No Save/Restore

The game runs but has no checkpointing. When the hero dies, the game is over with no way to recover.

### Implementation: Memento Pattern

A `Memento` captures game state snapshots. The `Caretaker` stores them. The `Originator` restores from a memento on hero death.
