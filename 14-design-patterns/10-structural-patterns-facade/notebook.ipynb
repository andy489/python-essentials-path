# Facade Pattern

## Overview

**Category:** Structural

The Facade pattern provides a **simplified interface** to a complex subsystem. It doesn't
hide the subsystem — clients can still access it directly — but the facade makes common
use cases dramatically simpler by grouping the right calls in the right order.

## Key Concepts

### The Problem
Accessing a database involves multiple steps:
1. Ask which database to use
2. Import the right Python module
3. Instantiate a control object
4. Build a connection string
5. Connect to the database
6. Run a query and parse results

Clients shouldn't have to know all of this.

### The Solution: Facade
A single `GetEmployeesFacade` class encapsulates all the above. The client calls one
method and gets back employee records.

### Structure
- **Facade** — exposes a simple, unified interface
- **Subsystem classes** — the complex components the facade orchestrates
- **Facade Factory** (optional) — selects the right concrete facade at runtime

### When to Use
- To provide a simple interface to a complex library or framework
- To layer your subsystems (use facades as entry points between layers)
- When you want to decouple client code from implementation details

## Demos
