# Interpreter Pattern

## Overview

**Category:** Behavioral

The Interpreter pattern defines a **grammar for a language** and provides an interpreter
to process sentences in that language. It is used to build simple domain-specific
languages (DSLs) where the grammar can be expressed as a class hierarchy.

## Key Concepts

- **Abstract Expression** — declares an `interpret(context)` interface
- **Terminal Expression** — interprets a terminal symbol in the grammar
- **Non-terminal Expression** — implements grammar rules involving other expressions
- **Context** — contains information global to the interpreter
- **AST (Abstract Syntax Tree)** — the tree of expression objects that represents a sentence

### Domain-Specific Languages
DSLs are languages designed for a specific problem domain:
- SQL (database queries)
- CSS (styling)
- HTML (markup)
- Regular expressions
- Python format strings: `[[fill]align][sign][#][0][width][.precision][type]`

### Backus-Naur Form (BNF)
Grammars are typically described in BNF notation. Example for a scrambled eggs recipe DSL:

```
expression ::= command | sequence | repetition
sequence   ::= expression ";" expression
command    ::= "break egg" | "mix in bowl" | "melt butter in pan" | "cook eggs"
repetition ::= while expression
```

### Consequences

**Benefits:**
- Easy to extend the grammar with new rules
- Easy to change how expressions are processed
- Can combine expressions freely

**Drawbacks:**
- Complex grammars become hard to maintain (class per rule)
- Consider parser generators (ANTLR, PLY) for complex languages
