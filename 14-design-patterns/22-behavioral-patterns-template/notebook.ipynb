# Template Method Pattern

## Overview

**Category:** Behavioral

The Template Method pattern defines the **skeleton of an algorithm** in a base class,
deferring some steps to subclasses. Subclasses can override specific steps without
changing the overall structure or step order. This is the classic example of the
Hollywood Principle: *"Don't call us, we'll call you."*

## Key Concepts

- **Template Method** — the method in the ABC that defines the algorithm's step order; usually `final` (not overridable)
- **Abstract Steps** — steps that *must* be overridden by subclasses
- **Concrete Steps** — default steps in the ABC that subclasses *may* override
- **Hooks** — optional override points with default no-op implementations

### Motivating Example: Travel
Whether you take a bus or a plane to your destination:
1. Pack your bags
2. Get to the departure point
3. Board the vehicle
4. Travel
5. Arrive and unpack

The *order* of steps is always the same. *How* each step works differs per transport type.

### Inversion of Control
The base class calls the subclass's overridden methods — not the other way around.
This is the "Hollywood Principle" in action.

### Consequences
- Maximises code reuse — common step order lives in one place
- Easy to extend — add new variants by subclassing
- Can make code harder to follow if the hierarchy is deep
