# Visitor Pattern

## Overview

**Category:** Behavioral

The Visitor pattern lets you **add new operations to an object structure without modifying
its classes**. A visitor object carries the new operation logic and is dispatched to the
right method based on the type of each element it visits.

## Key Concepts

- **Visitor** (`AbsVisitor`) — declares a `visit()` method for each type of element
- **Concrete Visitor** — implements one operation across all element types
- **Element** (`AbsTree`) — declares an `accept(visitor)` method
- **Concrete Element** — calls `visitor.visit(self)` inside `accept()`
- **Object Structure** — the collection of elements the visitor traverses

### Analogy
Think of a visitor to your house, a library, or a museum. The visitor brings a bag
(their tools/capabilities) and accesses the contents of each place they visit. Different
visitors can be sent to the same places to do different things.

### When to Use
- When you need to perform many distinct and unrelated operations on an object structure
- When the object structure rarely changes but you add new operations frequently

### Consequences
- Easy to add new operations (new visitor class)
- Harder to add new element types (all visitors must be updated)
- Can break encapsulation (visitor needs access to element internals)
- Works across class hierarchies and can accumulate state across visits

### Python Alternative
In Python, **class decorators** can sometimes replace the Visitor pattern for simpler cases.

## Demos

### Before: Operations Embedded in Tree Classes

Pretty-printing and searching logic are mixed directly into the tree node classes. Adding a new operation requires editing every node class.

### Implementation: Visitor Pattern

A `PrettyPrintVisitor` and `GetOldestVisitor` traverse the family tree independently. The tree structure is unchanged — new operations are new visitor classes.
