# Prototype Pattern

## Overview

**Category:** Creational

The Prototype pattern creates new objects by *cloning* an existing instance (the prototype)
rather than constructing them from scratch. This reduces class proliferation and supports
run-time inheritance — a clone starts as an exact copy and can then be customised.

## Key Concepts

- **Prototype** — the object being cloned, typically implements a `clone()` method
- **Shallow copy** — new object, but shared references to nested objects (`copy.copy()`)
- **Deep copy** — new object with recursively copied nested objects (`copy.deepcopy()`)
- **Prototype Manager** — a registry that stores named prototypes for lookup and cloning

### Motivating Example: Computer Models
Different computer models (Laptop, Tower, All-in-one) share many components. Instead of
building each from scratch, clone a prototype and change only what differs.

### Shallow vs Deep Copy

| | Shallow Copy | Deep Copy |
|---|---|---|
| New top-level object | ✓ | ✓ |
| Nested objects copied | ✗ (shared refs) | ✓ (recursive) |
| Risk | Mutations propagate | Possible infinite recursion |
| Python | `copy.copy()` | `copy.deepcopy()` |

### When to Use
- When object creation is expensive and a similar object already exists
- When classes to instantiate are specified at runtime
- When you want to avoid building a class hierarchy of factories

## Demos

### Demo: Shallow Copy

Clone a computer prototype using `copy.copy()`. Observe that nested mutable objects are shared between the original and the clone.

### Demo: Deep Copy

Use `copy.deepcopy()` to produce fully independent clones where nested objects are also duplicated.

### Demo: Prototype Manager

A central registry stores named prototype instances. Clients request a clone by name — no knowledge of concrete classes needed.
