# Singleton Pattern

## Overview

**Category:** Creational

The Singleton pattern ensures a class has **only one instance** and provides a global
point of access to it. It is commonly applied to resources that should be shared and
controlled — loggers, connection pools, device handles, configuration objects.

## Key Concepts

- One instance is created lazily or eagerly and cached
- All subsequent calls return the same instance
- The class controls its own instantiation

### Common Use Cases
- Device access (only one handle to a printer, serial port)
- Buffer and connection pools
- Web/database connection management
- Application-wide loggers

### Three Python Approaches

| Approach | Mechanism | Trade-offs |
|---|---|---|
| **Base class** | `__new__` returns cached instance | Simple; inheritance can be tricky |
| **Metaclass** | `__call__` intercepted on the metaclass | Clean; requires understanding metaclasses |
| **MonoState** | All instances share the same `__dict__` | Multiple instances, shared state; different intent |

### MonoState vs True Singleton
MonoState allows multiple instances but forces them all to share the same state. It is
sometimes preferred because it does not break inheritance and plays well with Python's
data model.

## Demos

### Demo: Classic Singleton via Base Class

Override `__new__` to cache the first instance and return it on all subsequent calls.

### Demo: Singleton Using a Base Class

Inherit singleton behaviour from a reusable `SingletonBase` so any subclass automatically becomes a singleton.

### Demo: Singleton via Metaclass

Implement the singleton logic in a metaclass `__call__`. The cleanest separation: the class itself is unaware of the pattern.

### Demo: MonoState Pattern

All instances share a class-level `__dict__`. Not a true singleton, but all objects behave identically — useful when you need multiple handles to the same shared state.
