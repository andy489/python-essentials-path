# Flyweight Pattern

## Overview

**Category:** Structural

The Flyweight pattern minimises memory use by sharing as much data as possible between
similar objects. Instead of each object storing all its state independently, intrinsic
(shared, invariant) state is centralised in a shared flyweight object, while extrinsic
(context-specific) state is passed in at runtime.

## Key Concepts

- **Intrinsic state** — stored in the flyweight; shared across many instances; immutable
- **Extrinsic state** — passed by the client at call time; varies per use; not stored
- **Flyweight Factory** — manages the pool of flyweight instances; returns existing ones

### Motivating Example: CERN's Large Hadron Collider
The LHC generates **billions of collision events per second**. Each event produces
thousands of particles. Creating a full Python object for every particle measurement would
be prohibitive. The Flyweight pattern centralises shared particle properties in a few
objects, passing per-event data externally.

### Structural Classification
- **Centralise state** for many objects
- **Reduce required object instances** dramatically
- All instances are controlled the same way — single instances are no longer independent

### When to Use
- When a large number of similar objects consume too much memory
- When most object state can be made extrinsic
- Common in game engines, particle systems, text rendering, map tiles
