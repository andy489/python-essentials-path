### Slide 12

Working with Breakpoints
You won’t always know where
breakpoints are needed before
debugging
Using the pdb debugger, you can
also interact with breakpoints


### Slide 13

break (alias b) – display all current breakpoints
Pass a line number or function name to break to set a breakpoint
Prefix the line number of function name with a file name to set a
breakpoint in a different file
Pass the break command a Python conditional as the second
argument to pause only if the conditional evaluates to True
clear (alias cl) – remove one or more breakpoints
Debugger Commands for Breakpoints


### Slide 14

step (alias s) – call a method or function on the current line and
then pause on the method or function being called (“step into”)
up (alias u) – return to the line of the most recent step command
(“step out”)
The stack trace is like a breadcrumb menu showing where the
current position is in the hierarchy of calls
down (alias d) – move one or more levels down in the stack trace
where (alias w) – show the current stack trace
Debugger Commands for Navigation
